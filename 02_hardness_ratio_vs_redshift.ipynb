```python
import pandas as pd
import numpy as np
from astropy.io import fits
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d

fits_file = 'chandra_COSMOS_legacy_opt_NIR_counterparts_20160113_4d.fits'
hdul = fits.open(fits_file)
df = pd.DataFrame(hdul[1].data)

df.replace(-99.000, np.nan, inplace=True)
df.replace(-99, np.nan, inplace=True)

# Filter for valid spectroscopic redshifts (z_spec)
mask_valid_z = (df['z_spec'].notna()) # & (df['z_spec'] > 0) # Add (df['Qg'] >= 1.5))
df_z = df[mask_valid_z].copy()

print(f"Total sources with valid z_spec: {len(df_z)}")

# Show all columns
pd.set_option('display.max_columns', None)
df.head(3)
```

## Two Selection Criteria

```python
pre_df = df_z.copy()

pre_df['flux_S'] = pd.to_numeric(pre_df['flux_S'], errors='coerce')
pre_df['i_mag'] = pd.to_numeric(pre_df['i_mag'], errors='coerce')

mask_valid = (pre_df['flux_S'] > 0) & (pre_df['i_mag'] > 0)
pre_df = pre_df[mask_valid].copy()

print(f"Sources with valid flux_S and i_mag: {len(pre_df)}")

pre_df['X_over_O'] = np.log10(pre_df['flux_S']) + (pre_df['i_mag'] / 2.5) + 5.352

# Two AGN criteria
# Criterion 1: L_0.5-10 keV > 3e42 erg/s (log10 > 42.5)
mask_luminosity = pre_df['Lx_0510'] > 42.5   # Where 42.5 = np.log10(3) + 42

# Criterion 2: -1 <= X/O <= 1
mask_xo = (pre_df['X_over_O'] >= -1.0) & (pre_df['X_over_O'] <= 1.0)

# Combine with OR
agn_mask = mask_luminosity | mask_xo

f1_sources = pre_df[agn_mask].copy()

print(f"Sources passing at least one criterion: {len(f1_sources)}")
print(f"Sources rejected: {len(pre_df) - len(f1_sources)}")
```

```python
# --- Ensure spec_type_label exists in both DataFrames ---
if 'spec_type_label' not in pre_df.columns:
    pre_df['spec_type_label'] = pre_df['spec_type'].apply(get_spec_type_label)

if 'spec_type_label' not in f1_sources.columns:
    f1_sources['spec_type_label'] = f1_sources['spec_type'].apply(get_spec_type_label)

# --- Define masks on the full pre-selected sample (pre_df) ---
mask_lum_pre = pre_df['Lx_0510'] > 42.5
mask_xo_pre = (pre_df['X_over_O'] >= -1.0) & (pre_df['X_over_O'] <= 1.0)

# --- Function to count criteria for a given label ---
def count_criteria(label):
    # Subset from pre_df (full sample)
    subset = pre_df[pre_df['spec_type_label'] == label]
    if len(subset) == 0:
        return {'Only Luminosity': 0, 'Only X/O': 0, 'Both': 0, 'Rejected': 0, 'Total': 0}
    
    # Masks for this subset
    lum = mask_lum_pre.loc[subset.index]
    xo = mask_xo_pre.loc[subset.index]
    
    only_lum = (lum & ~xo).sum()
    only_xo = (~lum & xo).sum()
    both = (lum & xo).sum()
    total = len(subset)
    
    # Rejected = total - (only_lum + only_xo + both)
    rejected = total - (only_lum + only_xo + both)
    
    return {'Total': total, 'Only Luminosity': only_lum, 'Only X/O': only_xo, 'Both': both, 
            'Rejected': rejected}

# --- Define the labels of interest ---
labels = ['BL AGN', 'Non-BL AGN', 'Star', 'Unassigned']
results = {label: count_criteria(label) for label in labels}

# --- Create a pandas DataFrame ---
df_counts = pd.DataFrame(results).T

# Add a row for "All" (sum of all categories)
all_counts = df_counts.sum()
df_counts.loc['All'] = all_counts

print(df_counts)
```

## HR vs z_spec (References at the very bottom)

### In The Beninggin

```python
# Helper to label spectral types
def get_spec_type_label(spec_type):
    if spec_type == 0:
        return 'Star'
    elif spec_type == 1:
        return 'BL AGN'
    elif spec_type == 2:
        return 'Non-BL AGN'
    else:
        return 'Unassigned'

# Galactic column and analytic HR model (exactly as in the first plot)
NH_GAL = 2.6e20
"""
Found in C1 Differential Sky coverage in Appendix C: Bias Corrections from Lanzuisi et al. (2018)
where they say they use it from Civano et al. (2016)
"""

def hr_curve_analytic(nh_cm2, z_array, gamma=1.7, # As Dr. Itsna suggest
                      e_soft=(0.5, 2.0), e_hard=(2.0, 7.0),
                      nh_gal=NH_GAL):
    """
    Approximate HR(z) for a power-law spectrum with intrinsic absorption.
    Uses a power-law approximation to the Morrison & McCammon (1983)
    photoelectric cross-section, evaluated in the REST frame. Overall, it is:
    F(E_obs) proportional to E_obs^(-Gamma) * exp[-1 * sigma{E_obs(1 + z)} * (N_H,gal + N_H,int)]
    """
    e  = np.logspace(np.log10(0.3), np.log10(10.0), 1000)
    de = np.gradient(e)

    def powerlaw(e_obs):
        return e_obs ** (-gamma)

    def transmission(e_obs, z, nh_int):
        e_rest = e_obs * (1.0 + z)
        sigma  = 2.4e-22 * (e_rest ** (-8.0 / 3.0))
        """
        2.4e-22 being the cm^2 per H atom at 1 keV of photon energy according to Table 2 from
        Morrison & McCammon (1983).
        
        Initially, I wanted to use -3 where it comes directly from the Note part of 
        aforementioned Table 2. 
        
        However, according to F et al. (1997), this is only the limiting case of having zero 
        metallicity. This means that -3.0 value is only valid if the gas is composed of purely 
        hydrogen and helium with no metals. The same paper later state that 'X-rays incoming on 
        gas with interstellar composition ... E^(-8/3)' and hence I use -8/3 instead.
        """
        tau    = sigma * (nh_gal + nh_int)
        return np.exp(-tau)

    def band_flux(e_min, e_max, z, nh_int):
        mask = (e >= e_min) & (e <= e_max)
        integrand = powerlaw(e[mask]) * transmission(e[mask], z, nh_int)
        return np.sum(integrand * de[mask])

    S = np.array([band_flux(*e_soft, z, nh_cm2) for z in z_array])
    H = np.array([band_flux(*e_hard, z, nh_cm2) for z in z_array])

    with np.errstate(divide='ignore', invalid='ignore'):
        hr = np.where((H + S) > 0, (H - S) / (H + S), 0.0)

    return hr

# Wrapper for single redshift (used in the summary)
def hr_theory(nh_cm2, z):
    return float(hr_curve_analytic(nh_cm2, np.array([z]))[0])
```

### First Plot (No Error Bars)

```python
# ── Prepare observed sample ──────────────────────────────────────────────────
obs = f1_sources.dropna(subset=['HR', 'z_spec']).copy()
print(f"AGN-selected sources with valid HR and z_spec: {len(obs)}")

if 'spec_type_label' not in obs.columns:
    obs['spec_type_label'] = obs['spec_type'].apply(get_spec_type_label)

# ── Generate curves ──────────────────────────────────────────────────────────
z_grid = np.linspace(0.01, 5.0, 200)

nh_values  = [1e22, 1e23, 1e24]                 # cm^-2
nh_labels  = [r'$N_H = 10^{22}$ cm$^{-2}$',
              r'$N_H = 10^{23}$ cm$^{-2}$',
              r'$N_H = 10^{24}$ cm$^{-2}$']
nh_colors  = ['red', 'grey', 'black']
nh_styles  = ['-', '--', ':']

curves = {lab: hr_curve_analytic(nh, z_grid)
          for nh, lab in zip(nh_values, nh_labels)}

# ── Plot ─────────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(11, 7))

spec_colors = {
    'Star':       'orange',
    'BL AGN':     'blue',
    'Non-BL AGN': 'green',
    'Unassigned': 'gray',
}
spec_markers = {
    'Star':       's',
    'BL AGN':     'o',
    'Non-BL AGN': 'o',
    'Unassigned': 'o',
}

for stype in ['Star', 'BL AGN', 'Non-BL AGN', 'Unassigned']:
    subset = obs[obs['spec_type_label'] == stype]
    if len(subset) == 0:
        continue
    ax.scatter(
        subset['z_spec'],
        subset['HR'],
        s=22 if stype != 'Unassigned' else 10,
        alpha=0.65 if stype != 'Unassigned' else 0.3,
        color=spec_colors[stype],
        marker=spec_markers[stype],
        edgecolors='none',
        label=f'{stype}',
    )

for lab, col, ls in zip(nh_labels, nh_colors, nh_styles):
    ax.plot(z_grid, curves[lab], color=col, linestyle=ls,
            linewidth=2, label=lab, zorder=5)

ax.set_xlabel(r'$z_{\rm spec}$', fontsize=14)
ax.set_ylabel(r'Hardness Ratio $(H-S)/(H+S)$', fontsize=14)
ax.set_title('Hardness Ratio vs. Spectroscopic Redshift\n'
             r'(AGN-selected sample, $\Gamma = 1.7$ curves overlaid)',
             fontsize=15)
ax.legend(loc='lower left', fontsize=10, framealpha=0.9)
ax.grid(True, alpha=0.3)
ax.set_xlim(0, 5)
ax.set_ylim(-1, 1)

plt.tight_layout()
plt.show()
```

### Summary

```python
# ── Prepare observed sample (same as above) ─────────────────────────────────
obs = f1_sources.dropna(subset=['HR', 'z_spec']).copy()
if 'spec_type_label' not in obs.columns:
    obs['spec_type_label'] = obs['spec_type'].apply(get_spec_type_label)

print(f"AGN-selected sources with valid HR and z_spec: {len(obs)}")

# ── Invert HR → log N_H at the source redshift (same model) ─────────────────
logNH_grid = np.arange(20.0, 25.5, 0.05)
NH_grid    = 10.0 ** logNH_grid          # cm^-2

def nh_from_hr(hr_obs, z_obs):
    hr_model = np.array([hr_theory(nh, z_obs) for nh in NH_grid])
    idx = np.argmin(np.abs(hr_model - hr_obs))   # HR is monotonic in N_H
    return logNH_grid[idx]

obs['logNH_from_HR'] = [
    nh_from_hr(hr, z) for hr, z in zip(obs['HR'].values, obs['z_spec'].values)
]

# ── Classify ────────────────────────────────────────────────────────────────
def classify(lognh):
    if   lognh < 22.0: return 'Unobscured'
    elif lognh < 24.0: return 'Obscured'
    else:              return 'Heavily obscured'

obs['obscuration_class'] = obs['logNH_from_HR'].apply(classify)

# ── Text summary ───────────────────────────────────────────────────────────
classes = ['Unobscured', 'Obscured', 'Heavily obscured']
cats    = ['Star', 'BL AGN', 'Non-BL AGN', 'Unassigned']

print("\n" + "=" * 80)
print("OBSCURATION SUMMARY  (N_H inferred from Hardness Ratio)")
print("  Unobscured        : log N_H <  22")
print("  Obscured          : 22 <= log N_H <  24")
print("  Heavily obscured  : log N_H >= 24")
print("=" * 80)

for cat in cats:
    sub = obs[obs['spec_type_label'] == cat]
    n = len(sub)
    print(f"\n{cat}  (N = {n})")
    print("-" * 80)
    if n == 0:
        continue
    for cls in classes:
        k   = (sub['obscuration_class'] == cls).sum()
        pct = k / n * 100.0
        print(f"  {cls:<20s} : {k:4d} / {n:4d}   ({pct:5.1f}%)")
    med = sub['logNH_from_HR'].median()
    print(f"  median log N_H       : {med:.2f}")

# Combined AGN row (BL + Non-BL)
agn = obs[obs['spec_type_label'].isin(['BL AGN', 'Non-BL AGN'])]
n   = len(agn)
print(f"\nAll AGN  (N = {n})")
print("-" * 80)
for cls in classes:
    k   = (agn['obscuration_class'] == cls).sum()
    pct = k / n * 100.0 if n else 0.0
    print(f"  {cls:<20s} : {k:4d} / {n:4d}   ({pct:5.1f}%)")
if n:
    print(f"  median log N_H       : {agn['logNH_from_HR'].median():.2f}")
print("=" * 80)
```
